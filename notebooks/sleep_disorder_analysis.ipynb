# Sleep Disorder Classification: Exploratory Data Analysis & Preprocessing

## 1. Project Overview
This notebook performs exploratory data analysis (EDA), data cleaning, feature engineering, and statistical analysis on the **Sleep Health and Lifestyle Dataset**.

### Objective:
Understand the relationship between lifestyle habits, physiological metrics, and sleep disorders (**None**, **Insomnia**, **Sleep Apnea**).


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
%matplotlib inline
```

## 2. Load Dataset
We load `sleep_health.csv` from the `data/` directory and inspect the structure.

```python
df = pd.read_csv('../data/sleep_health.csv')
print('Dataset shape:', df.shape)
df.head()
```

## 3. Data Cleaning and Handling Missing Values
- **Sleep Disorder**: Missing values (`NaN`) correspond to healthy individuals with **No Sleep Disorder** (`None`).
- **BMI Category**: Contains `'Normal Weight'` which is synonymous with `'Normal'`.
- **Blood Pressure**: Compound string format `'Systolic/Diastolic'` which we parse into two numerical features.

```python
# Clean target
df['Sleep Disorder'] = df['Sleep Disorder'].fillna('None').astype(str).str.strip()

# Standardize BMI category
df['BMI Category'] = df['BMI Category'].replace({'Normal Weight': 'Normal'})

# Parse Blood Pressure
bp_split = df['Blood Pressure'].str.split('/', expand=True)
df['Systolic_BP'] = pd.to_numeric(bp_split[0], errors='coerce')
df['Diastolic_BP'] = pd.to_numeric(bp_split[1], errors='coerce')

print('Missing values after cleaning:')
print(df.isnull().sum())
print('\nTarget distribution:')
print(df['Sleep Disorder'].value_counts())
```

## 4. Summary Statistics

```python
df.describe().T
```

## 5. Visualizations & Statistical Relationships

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=df, x='Sleep Disorder', ax=axes[0], palette={'None': '#2ecc71', 'Insomnia': '#e67e22', 'Sleep Apnea': '#e74c3c'}, order=['None', 'Insomnia', 'Sleep Apnea'])
axes[0].set_title('Target Class Distribution')

sns.countplot(data=df, x='BMI Category', ax=axes[1], palette='Blues_d', order=['Normal', 'Overweight', 'Obese'])
axes[1].set_title('BMI Category Distribution')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
palette = {'None': '#2ecc71', 'Insomnia': '#e67e22', 'Sleep Apnea': '#e74c3c'}
sns.boxplot(data=df, x='Sleep Disorder', y='Sleep Duration', ax=axes[0], palette=palette, order=['None', 'Insomnia', 'Sleep Apnea'])
axes[0].set_title('Sleep Duration vs Disorder')

sns.boxplot(data=df, x='Sleep Disorder', y='Quality of Sleep', ax=axes[1], palette=palette, order=['None', 'Insomnia', 'Sleep Apnea'])
axes[1].set_title('Quality of Sleep vs Disorder')

sns.boxplot(data=df, x='Sleep Disorder', y='Stress Level', ax=axes[2], palette=palette, order=['None', 'Insomnia', 'Sleep Apnea'])
axes[2].set_title('Stress Level vs Disorder')
plt.tight_layout()
plt.show()
```

```python
num_cols = ['Age', 'Sleep Duration', 'Quality of Sleep', 'Physical Activity Level', 'Stress Level', 'Heart Rate', 'Daily Steps', 'Systolic_BP', 'Diastolic_BP']
plt.figure(figsize=(10, 8))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Heatmap')
plt.show()
```

## 6. Key Findings
1. **Sleep Duration & Quality**: Healthy subjects average > 7.5 hours of sleep with quality ratings >= 7. Insomnia sufferers sleep under 6.5 hours with significantly higher stress levels.
2. **BMI & Blood Pressure**: Sleep Apnea strongly correlates with Overweight/Obese status and higher blood pressure readings (systolic > 130 mmHg).
3. **Physical Activity**: Regular physical activity and daily step counts correlate positively with sleep quality and negatively with stress levels.
